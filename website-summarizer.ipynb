```python
import os
from IPython.display import Markdown, display
from dotenv import load_dotenv
from openai import OpenAI
from scraper import fetch_web_content
```

```python
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    raise RuntimeError("OPENAI_API_KEY is not set in the environment")
```

```python
openai = OpenAI()
```

```python
system_prompt = """ You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be naviation related.
Respond in markdown. Do not wrap the markdown in a code block - respond with just markdown """
```

```python
user_prompt_prefix = """ 
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too
"""
```

```python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

```python
def summarize(url):
    website = fetch_web_content(url)
    response = openai.chat.completions.create(
        model='gpt-4.1-mini',
        messages=messages_for(website)
    )
    return response.choices[0].message.content
```

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary('https://www.apisec.ai/')
```
