```python
import os
import requests
from dotenv import load_dotenv
from openai.types.responses import ResponseTextDeltaEvent
from agents import Agent, Runner, trace, function_tool, SQLiteSession
load_dotenv(override=True)
```

```python
agent = Agent(name="Assistant", model="gpt-5.4-mini")
```

```python
response = await Runner.run(agent, "Hi there my name is Dennis")
print(response.final_output)
```

```python
reponse = await Runner.run(agent, "What is my name?")
print(response.final_output)
```

```python
convo_history = response.to_input_list() + [{"role": "user", "content": "What's my name?"}]
```

```python
# Runner.run can accept a user prompt or a list of prompts, so we provide the convo history as input to create the illusion of memory
res = await Runner.run(agent, convo_history)
print(res.final_output)
```
